```python
# Data manipulation
import pandas as pd
import numpy as np

# Visualisation
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', '{:.2f}'.format)

# Visualisation settings
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
import os

# Project data directory
DATA_PATH = "../data"

# Load datasets
ball = pd.read_csv(
    os.path.join(DATA_PATH, "ball_by_ball.csv")
)

matches = pd.read_csv(
    os.path.join(DATA_PATH, "matches.csv")
)

competitions = pd.read_csv(
    os.path.join(DATA_PATH, "competitions.csv")
)

teams = pd.read_csv(
    os.path.join(DATA_PATH, "teams.csv")
)

playing_eleven = pd.read_csv(
    os.path.join(DATA_PATH, "playing_eleven.csv")
)

player_mapping = pd.read_csv(
    os.path.join(DATA_PATH, "competition_team_player_mapping.csv")
)

print("All datasets loaded successfully!")
```

```python
datasets = {
    "Ball by Ball": ball,
    "Matches": matches,
    "Competitions": competitions,
    "Teams": teams,
    "Playing Eleven": playing_eleven,
    "Player Mapping": player_mapping
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

```python
# Checking missing values
for name, df in datasets.items():

    print(f"\n{name}")

    missing = pd.DataFrame({
        "Missing Values": df.isnull().sum(),
        "Missing Percentage": (
            df.isnull().mean() * 100
        ).round(2)
    })

    print(missing[missing["Missing Values"] > 0])
```

```python
ball_matches = ball.merge(
    matches[
        [
            "match_id",
            "comp_id",
            "match_date",
            "venue_country",
            "home_team",
            "away_team",
            "revised_target",
            "first_inn_revised_overs",
            "second_inn_revised_overs"
        ]
    ],
    on="match_id",
    how="left"
)

ball_matches = ball_matches.merge(
    competitions[
        ["comp_id", "comp_name", "match_format", "season"]
    ],
    on="comp_id",
    how="left"
)

print("Merged dataset shape:", ball_matches.shape)

display(
    ball_matches[
        [
            "match_id",
            "match_date",
            "comp_name",
            "match_format",
            "season",
            "innings",
            "striker_name",
            "bowler_name",
            "runs"
        ]
    ].head()
)
```

```python
print(ball["bowl_speed"].value_counts().head(15))

print("Negative values:",
      (ball["bowl_speed"] < 0).sum())

print("Zero values:",
      (ball["bowl_speed"] == 0).sum())
```

```python
# Merge ball-by-ball data with match information

ball_matches = ball.merge(
    matches,
    on="match_id",
    how="left",
    validate="many_to_one"
)

# Add competition information

ball_matches = ball_matches.merge(
    competitions,
    on="comp_id",
    how="left",
    validate="many_to_one"
)

print("Merged dataset shape:", ball_matches.shape)

display(ball_matches.head())
```

```python
print("Missing competition names:",
      ball_matches["comp_name"].isna().sum())

print("Missing match dates:",
      ball_matches["match_date"].isna().sum())

print("Unique matches:",
      ball_matches["match_id"].nunique())
```

```python
print("Unmatched match records:",
      ball_matches["comp_id"].isna().sum())

print("Unmatched competition records:",
      ball_matches["comp_name"].isna().sum())

print("Unique match IDs:",
      ball_matches["match_id"].nunique())

print("Unique competition IDs:",
      ball_matches["comp_id"].nunique())
```

```python
extra_innings = ball_matches[
    ball_matches["innings"] > 2
]

display(
    extra_innings[
        [
            "match_id",
            "comp_name",
            "match_date",
            "innings",
            "striker_name",
            "bowler_name",
            "runs"
        ]
    ]
)
```

```python
extra_match_ids = (
    ball_matches.loc[
        ball_matches["innings"] > 2,
        "match_id"
    ].unique()
)

display(
    matches[
        matches["match_id"].isin(extra_match_ids)
    ]
)
```

```python
print("Available extras-related columns:")

print([
    col for col in ball.columns
    if "extra" in col.lower() or "run" in col.lower()
])
```

```python
print("Extras type distribution:")

display(
    ball["extras_type"]
    .value_counts(dropna=False)
)

print("Runs distribution by extras type:")

display(
    ball.groupby(
        "extras_type",
        dropna=False
    )["runs"]
    .agg(["count", "sum", "mean"])
    .sort_values("count", ascending=False)
)
```

```python
# Examine extras-related columns

extras_columns = [
    col for col in ball.columns
    if "extra" in col.lower() or "run" in col.lower()
]

print(extras_columns)
```

```python
# Distribution of extras

display(
    ball["extras_type"]
    .value_counts(dropna=False)
)
```

```python
# Runs distribution by extras type

extras_analysis = (
    ball.groupby("extras_type", dropna=False)
    .agg(
        deliveries=("match_id", "count"),
        runs=("runs", "sum"),
        extras1=("extras1_runs", "sum"),
        extras2=("extras_2_runs", "sum")
    )
    .reset_index()
)

display(extras_analysis)
```
